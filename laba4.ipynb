```python
# Узел для связного списка
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

# номер 1
class StackArray:
    # стек на массиве
    def __init__(self):
        self.arr = []

    def push(self, x):        # добавить элемент
        self.arr.append(x)

    def pop(self):            # удалить верхний элемент
        if self.is_empty():
            return None
        return self.arr.pop()

    def top(self):            # посмотреть верхний элемент
        return self.arr[-1]

    def is_empty(self):       # проверить пуст ли стек
        return len(self.arr) == 0


class StackList:
    # стек на связном списке
    def __init__(self):
        self.head = None

    def push(self, x):
        node = Node(x)
        node.next = self.head
        self.head = node

    def pop(self):
        if self.is_empty():
            return None
        x = self.head.value
        self.head = self.head.next
        return x

    def top(self):
        return self.head.value

    def is_empty(self):
        return self.head is None


class QueueArray:
    # очередь на массиве
    def __init__(self):
        self.arr = []

    def push(self, x):        # добавить в конец
        self.arr.append(x)

    def pop(self):            # удалить из начала
        if self.is_empty():
            return None
        return self.arr.pop(0)

    def is_empty(self):
        return len(self.arr) == 0


class QueueList:
    # очередь на связном списке
    def __init__(self):
        self.first = None
        self.last = None

    def push(self, x):
        node = Node(x)
        if self.last:
            self.last.next = node
        else:
            self.first = node
        self.last = node

    def pop(self):
        if self.is_empty():
            return None
        x = self.first.value
        self.first = self.first.next
        if self.first is None:
            self.last = None
        return x

    def is_empty(self):
        return self.first is None


class DequeArray:
    # дек на массиве
    def __init__(self):
        self.arr = []

    def push_front(self, x):  # добавить в начало
        self.arr.insert(0, x)

    def push_back(self, x):   # добавить в конец
        self.arr.append(x)

    def pop_front(self):      # удалить из начала
        return self.arr.pop(0)

    def pop_back(self):       # удалить из конца
        return self.arr.pop()

    def is_empty(self):
        return len(self.arr) == 0


class DequeList:
    # дек на связном списке
    def __init__(self):
        self.first = None
        self.last = None

    def push_front(self, x):
        node = Node(x)
        node.next = self.first
        self.first = node
        if self.last is None:
            self.last = node

    def push_back(self, x):
        node = Node(x)
        if self.last:
            self.last.next = node
        else:
            self.first = node
        self.last = node

    def pop_front(self):
        x = self.first.value
        self.first = self.first.next
        if self.first is None:
            self.last = None
        return x

    def pop_back(self):
        if self.first == self.last:
            x = self.first.value
            self.first = self.last = None
            return x
        cur = self.first
        while cur.next != self.last:
            cur = cur.next
        x = self.last.value
        cur.next = None
        self.last = cur
        return x

    def is_empty(self):
        return self.first is None


# номер 2 проверка правильности скобок

def check_brackets(s):
    st = StackArray()
    pairs = {')': '(', ']': '[', '}': '{'}

    for c in s:
        if c in '([{':
            st.push(c)
        else:
            if st.is_empty() or st.pop() != pairs[c]:
                return False

    return st.is_empty()


# номер 3 вычисление постфиксного выражения

def calc_postfix(s):
    st = StackArray()

    for c in s.split():
        if c.isdigit():
            st.push(int(c))
        else:
            b = st.pop()
            a = st.pop()
            if c == '+':
                st.push(a + b)
            elif c == '-':
                st.push(a - b)
            elif c == '*':
                st.push(a * b)
            elif c == '/':
                st.push(a / b)

    return st.pop()


# номер 4 перевод из инфиксной в постфиксную форму

def infix_to_postfix(s):
    pr = {'+': 1, '-': 1, '*': 2, '/': 2}
    st = StackArray()
    res = []

    for c in s.split():
        if c.isdigit():
            res.append(c)
        elif c == '(':
            st.push(c)
        elif c == ')':
            while st.top() != '(':
                res.append(st.pop())
            st.pop()
        else:
            while (not st.is_empty() and
                   st.top() in pr and
                   pr[st.top()] >= pr[c]):
                res.append(st.pop())
            st.push(c)

    while not st.is_empty():
        res.append(st.pop())

    return ' '.join(res)

print(check_brackets("{[()]}"))
print(calc_postfix("2 3 + 4 *"))
print(infix_to_postfix("( 2 + 3 ) * 4"))
```
